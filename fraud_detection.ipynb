```python

import pandas as pd

# Sample transaction dataset
data = {
    "Transaction_ID": [101, 102, 103, 104, 105, 106],
    "Amount": [500, 1200, 95000, 700, 150000, 900],
    "Transactions_Per_Hour": [1, 2, 15, 1, 20, 2],
    "International": [0, 0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

# Simple risk assessment rules
def assess_risk(row):
    if row["Amount"] > 50000 or row["Transactions_Per_Hour"] > 10:
        return "High Risk"
    elif row["International"] == 1:
        return "Medium Risk"
    else:
        return "Low Risk"

df["Risk_Level"] = df.apply(assess_risk, axis=1)

print("FRAUD RISK ANALYSIS")
print(df.to_string(index=False))

print("\nRisk Summary:")
print(df["Risk_Level"].value_counts())
```
